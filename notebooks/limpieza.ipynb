```python
import pandas as pd

print(pd.__version__)
```

```python
df = pd.read_csv("../data/df1_actividad_clientes.csv")

df.head()
```

#hacemos un diagnostico previo a la limpieza

```python
#generamos un diagnostico para comprobar si hay fallos en el df

print("Filas y columnas:")
print(df.shape)

print("\nTipos de datos:")
print(df.dtypes)

print("\nValores nulos:")
print(df.isnull().sum())

print("\nDuplicados:")
print(df.duplicated().sum())
```

```python
nulos = df.isnull().sum()

nulos[nulos > 0].sort_values(ascending=False)
```

```python
print("Estados del pedido:")
print(df["order_status"].value_counts())

print("\nEstados:")
print(df["customer_state"].value_counts())

print("\nValoraciones:")
print(df["review_score"].value_counts().sort_index())
```

```python
columnas_fecha = [
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for columna in columnas_fecha:
    df[columna] = pd.to_datetime(df[columna], errors="coerce")
```

```python
df.dtypes
```

```python
filas = len(df)
pedidos_unicos = df["order_id"].nunique()

print("Filas:", filas)
print("Pedidos únicos:", pedidos_unicos)
print("¿El grano es correcto?", filas == pedidos_unicos)

# Comprobamos el grano del DataFrame:
# contamos el número total de filas y lo comparamos con el número
# de pedidos (order_id) únicos. Si ambos valores coinciden,
# significa que cada fila representa un único pedido.
```

```python
df.to_csv(
    "../data/df1_actividad_clientes_limpio.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Dataset limpio exportado correctamente.")
```
